# Research Objective

*Requires data modeling and quantitative research in Transport, Infrastructure & Logistics*

The goal of this research is to investigate the relationship between public transport (PT) accessibility and population density in the Netherlands, and to explore whether this relationship contributes to unequal accessibility between urban and rural populations. By mapping neighborhoods in which PT accessibility is relatively low given their population density, the research aims to provide policymakers and urban planners with useful information on areas that need attention in order for the sustainable mobility transition to be socially inclusive.

To do this, the research pursues the following objectives:

1. Develop a PT accessibility indicator at the neighborhood level, based on the number of stops within walking distance and the scheduled service frequency at these stops (GTFS).
2. Combine this indicator with CBS data on population density in one dataset covering Dutch neighborhoods.
3. Quantify the relationship between population density and PT accessibility using correlation and regression analysis.
4. Compare this relationship between urban and rural neighborhoods.
5. Identify and map neighborhoods in which PT accessibility deviates substantially from what would be expected based on their population density.

# Research Questions

**Main research question:**

To what extent is population density related to PT accessibility at the neighborhood level in the Netherlands, and does this relationship differ between urban and rural areas?

# Introduction

As the Dutch mobility system becomes increasingly sustainable, concerns are growing that lower-income households may be unable to afford the transition. In particular, policies that increase the cost of fossil-fuelled car use could disproportionately affect households that are highly car-dependent but lack the financial means to adopt alternatives such as electric vehicles (Dalla Longa et al., 2024). Public transport could provide a more sustainable and inclusive alternative. However, it can only perform this role if services are affordable, reliable and  accessible.

The availability of public transport is not distributed equally across the Netherlands. Public transport accessibility is generally higher in urban areas, whereas residents in rural areas often face lower frequencies and fewer connections. This is expected to be closely related to population density: densely populated areas create a larger number of potential passengers, making frequent public transport more economically attractive, while spatially dispersed populations are more difficult and costly to serve. As a result, lower-income households living in low-density areas may remain dependent on private cars even when rising costs make car use increasingly unaffordable.

This creates a potential mismatch between sustainability and social equity. Although the transition aims to reduce emissions, it may increase the risk of transport poverty if accessible and affordable alternatives are unavailable. Dalla Longa et al. (2024) find that transport-poverty risks are proportionally highest in rural areas. Examining the relationship between population density and public transport provision is therefore essential for understanding whether the transition towards sustainable mobility is also socially inclusive. Secondly, identifying the potential need for additional policy could be helpful information for policymakers and urban planners to make up for these differences.

This study therefore examines how population density is related to public transport provision and whether this relationship may contribute to unequal accessibility between urban and rural populations.



# Data Pipeline

This analysis will combine population and public transport data to make a clear representation of how public transport provision varies across neighborhoods in the Netherlands. Population density and neighborhood boundaries will first be obtained from the website of the CBS. GTFS timetable data will be processed to determine the locations of public transport stops and the frequency of scheduled services at each stop during a representative period.

The two datasets will then be combined using geopandas. For each neighborhood, public transport stops within a predefined distance will be identified. Based on the data from the timetable and the number of stops within the walking distance, a public transport accessibility indicator will be calculated.

This produces a dataset containing population density and public transport accessibility at a neighborhood level. Correlation and regression analyses will then be used to show the relationship between these two variables.

To make our analyses more accurate, neighborhoods will also get a label as either urban or rural. This makes it possible to compare the relationship between population density and public transport accessibility between different region types.

Finally, the results will be visualized using scatter plots and maps. These will show the spatial distribution of public transport accessibility and its relationship with population density.

# Scale of the Study

The geographical scope of this study covers the Netherlands at the neighborhood level (*CBS buurten*). Analyzing accessibility at this spatial resolution is essential because public transport service catchments operate at a localized walking scale, typically defined by CROW guidelines as a walking distance of 400 to 800 meters to local bus and tram stops. Aggregating the analysis to larger spatial units, such as municipalities, would obscure localized service gaps and conceal pockets of transport poverty. To avoid statistical distortion from unpopulated rural land, water bodies, and industrial sites, neighborhoods with fewer than 50 registered residents are excluded from the dataset. Neighborhoods are categorized into urban and rural environments following the official CBS urbanity classification (*stedelijkheid*), which classifies areas based on surrounding address density (*omgevingsadressendichtheid*).

The temporal scope is defined by a cross-sectional snapshot of a single representative working weekday (a Tuesday or Thursday in October 2026) outside of school vacation periods and scheduled national rail disruptions. In transport planning methodology, mid-week working days represent the structural baseline of public transport supply, avoiding the reduced frequencies of weekend timetables and Monday or Friday hybrid-work demand variations. The analysis focuses on scheduled departures between 07:00 and 19:00, capturing both peak commuting periods and baseline daytime frequencies. This timetable snapshot from the national GTFS feed is cross-referenced with the latest CBS *Kerncijfers Wijken en Buurten* demographic release. Limiting the study to a single representative 12-hour weekday ensures the data processing remains computationally feasible within Python and GeoPandas, while accurately capturing the structural public transport supply provided by transit agencies.

# Results & Empirical Findings

## 1. Spatial Distribution of Public Transport Accessibility (Sub-Question 1)
Across all analyzed Dutch neighborhoods ($N = 14,041$), public transport accessibility displays substantial spatial variation. The distribution of weekly scheduled departures within a 500-meter walking buffer is heavily right-skewed:
* **Mean supply:** 4,357 scheduled departures per week per neighborhood.
* **Median supply:** 2,115 departures per week, demonstrating that a minority of high-density transit corridors substantially elevate the national average.
* **Network gaps:** 704 neighborhoods record zero scheduled transit stops within their 500-meter walking boundary, located predominantly in peripheral agricultural regions and exterior industrial zones.

## 2. Quantitative Relationship between Density and Accessibility (Sub-Question 2)
To examine the relationship between neighborhood population density and scheduled public transport departures, both variables were log-transformed ($\log_{10}$) to stabilize variance and reflect diminishing marginal returns in transit provision.

The estimated Ordinary Least Squares (OLS) model is formulated as:
$$\log_{10}(\text{Departures}_{i} + 1) = 2.1147 + 0.3438 \cdot \log_{10}(\text{Density}_{i})$$

* **Sample size:** $N = 14,041$
* **Statistical significance:** $p < 10^{-15}$ (standard error of slope = $0.00815$)
* **Model explanatory power:** $R = 0.3354$, $R^2 = 0.1125$. Population density accounts for approximately 11.3% of the total variance in localized public transport supply.
* **Rank correlation:** The non-parametric Spearman rank correlation is $r_s = 0.4766$ ($p < 10^{-15}$), and the untransformed Pearson correlation is $r = 0.5091$, confirming a positive and statistically significant baseline association.

![Regression Scatter](Images/regression_scatter.png)

As demonstrated in the scatter plot, a quadratic log-log specification provides an improved fit ($R^2 = 0.150$). At lower densities ($< 10^2\text{ inh/km}^2$), service levels are largely governed by regional baseline connectivity mandates (e.g., regional bus corridors) rather than local passenger demand. Beyond $1,000\text{ inh/km}^2$, service frequencies scale rapidly with agglomeration density.

## 3. Urban versus Rural Disparities (Sub-Question 3)
Stratifying the data across the five official CBS urbanity tiers (*omgevingsadressendichtheid*) demonstrates that public transport provision is primarily driven by macro-level urban classification rather than fine-grained local density variations:

| CBS Urbanity Class | Sample ($N$) | Mean Density ($\text{inh/km}^2$) | Mean Departures/wk | Log-log Slope ($\beta_1$) | $R^2$ |
| :--- | :---: | :---: | :---: | :---: | :---: |
| **1: Very strongly urban** (*Zeer sterk stedelijk*) | 1,970 | 9,231.9 | 13,899.4 | +0.1621 | 0.0062 |
| **2: Strongly urban** (*Sterk stedelijk*) | 2,457 | 5,270.2 | 5,820.0 | -0.0877 | 0.0056 |
| **3: Moderately urban** (*Matig stedelijk*) | 1,902 | 3,746.7 | 3,500.2 | -0.1150 | 0.0152 |
| **4: Slightly urban** (*Weinig stedelijk*) | 2,111 | 2,261.6 | 2,529.3 | -0.1355 | 0.0220 |
| **5: Non-urban / Rural** (*Niet-stedelijk*) | 5,601 | 613.0 | 1,339.8 | -0.1301 | 0.0095 |

The urban-rural divide is pronounced: neighborhoods in Class 1 average more than ten times the weekly departure volume of non-urban areas (13,899 vs. 1,340 departures). Within classes 2 through 5, the internal regression slope is slightly negative ($\beta_1 \approx -0.09$ to $-0.14$). In suburban and rural municipalities, higher neighborhood density often corresponds to purely residential commuter enclaves situated away from main railway and bus corridors, whereas lower-density peripheral neighborhoods frequently intersect regional transit arteries.

## 4. Residual Identification & Spatial Outliers (Sub-Question 4)
Standardized regression residuals ($z$-scores) were computed to locate neighborhoods where public transport supply lags significantly behind model expectations given their density ($z < -1.0$):
* A total of 1,285 neighborhoods (9.15% of the sample) are classified as underserved relative to density.
* The lowest residual scores ($z \approx -3.76$ to $-3.92$) cluster in dense urban and suburban municipalities, including *Bossenbuurt hoog-West* and *Oosterheem-Centrum* (Zoetermeer), *Visserijbuurt* and *Statenkwartier* ('s-Gravenhage), and *Gillisbuurt* (Delft).

# Sensitivity Analysis: Catchment Buffer Distances
To test the robustness of the accessibility indicator, the spatial buffering process was evaluated for walking catchment thresholds of 300 meters and 800 meters alongside the 500-meter baseline.

![Sensitivity Lines](Images/sensitivity_regression_lines.png)

* **300-meter buffer:** Explains $R^2 = 0.09$ of the variance. At this tighter threshold, micro-level stop placement and local street morphology introduce higher stochastic variation.
* **500-meter buffer (Baseline):** Conforms to Dutch CROW walking guidelines while maintaining robust spatial coverage ($R^2 = 0.11$).
* **800-meter buffer:** Yields $R^2 = 0.13$, displaying a parallel upward shift in expected departure volumes as wider catchments capture multiple stops.

The consistency of the slope coefficients across all three thresholds confirms that the observed positive relationship between population density and service frequency is structurally robust to buffer specifications.

# Discussion & Limitations

### Data Pipeline Reflection and the 262-Neighborhood Anomaly
Cross-referencing `buurt_stop_matches.csv` ($53,291$ matches) with `stop_frequencies.csv` ($48,477$ active stops) revealed that 262 urban neighborhoods contain physical stops within 500 meters yet record zero weekly scheduled departures:
1. **Demand-Responsive Transit Filtering:** Flexible, on-demand services (*buurtbussen*, dial-a-ride, flexvervoer) frequently register placeholder departures in raw GTFS feeds. To prevent distortions in regular scheduled transit capacity, these services were filtered out during preprocessing.
2. **Buffer Edge Effects:** Neighborhoods located near stops served primarily by filtered flexible routes registered zero departures, even when high-frequency tram or bus corridors were located just outside the 500-meter catchment boundary.

### Transport Poverty and Policy Implications
Because dataset CBS 86358NED does not record neighborhood-level household income, the direct quantitative interaction between income, density, and public transport access could not be estimated parametrically. Nevertheless, the empirical findings align with the transport poverty framework described by Dalla Longa et al. (2024). In rural and low-density areas (Classes 4 and 5), where baseline transit provision averages fewer than 1,340 weekly departures, car dependency remains structurally high. Policies aimed at disincentivizing private vehicle use through taxation or environmental zones risk exacerbating transport poverty in peripheral communities unless accompanied by targeted public transit investments.

# Conclusion
This study evaluated the relationship between population density and public transport accessibility across Dutch neighborhoods and examined how this dynamic varies across the urban-rural continuum. 

The analysis reveals a statistically significant positive relationship ($\log_{10}(\text{Departures}+1) = 2.1147 + 0.3438 \cdot \log_{10}(\text{Density})$, $p < 10^{-15}$), with population density explaining 11.3% of the variance in local transit supply. However, this relationship differs substantially between urban and rural settings. While urban neighborhoods benefit from more than tenfold the departure frequency of rural areas, density variations *within* suburban and rural tiers do not correlate with improved transit supply. Transport planning in the Netherlands remains concentrated around metropolitan agglomerations, underscoring the need for targeted regional transit policies to ensure an equitable transition to sustainable mobility.
